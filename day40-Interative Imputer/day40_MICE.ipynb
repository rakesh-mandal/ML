```python
import numpy as np
import pandas as pd
# 1. You must explicitly enable IterativeImputer first
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
from sklearn.linear_model import LinearRegression
```

```python
df = np.round(pd.read_csv('50_Startups.csv')[['R&D Spend','Administration','Marketing Spend','Profit']]/10000)
np.random.seed(9)
df = df.sample(5)
df
```

```python
df = df.iloc[:,0:-1]
df
```

```python
df.iloc[1,0] = np.nan
df.iloc[3,1] = np.nan
df.iloc[-1,-1] = np.nan
```

```python
df.shape
```

```python
df
```

```python
# 2. Initialize the imputer matching your manual settings
imputer = IterativeImputer(
    estimator=LinearRegression(), # Switch from default BayesianRidge to match your notebook
    max_iter=2,                   # Number of iterations
    initial_strategy='mean',      # 0th iteration baseline
    imputation_order='roman',     # Process columns left-to-right index order
    random_state=9
)
```

```python
# 3. Fit and transform the data
imputed_array = imputer.fit_transform(df)
```

```python
imputed_array
```

```python
# Convert back to a readable DataFrame
df_imputed = pd.DataFrame(imputed_array, columns=df.columns)
print(np.round(df_imputed, 2))
```
